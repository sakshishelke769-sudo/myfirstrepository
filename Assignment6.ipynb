```python
# Analyze spending by gender 

gender_spending = df.groupby(
    "Gender"
)["Purchase_Value"].sum()
print(gender_spending)
```

```python
# Analyze spending by age group 

def age_group(age):
    if age <=25:
        return"18-25"
    elif age<=35:
        return"26-35"
    elif age<=45:
        return"36-45"
    else:
        return"46-55"
df["Age_Group"]=df["Age"].apply(age_group)

age_spending = df.groupby(
    "Age_Group"
)["Purchase_Value"].sum().sort_values(ascending=False)

print(age_spending)
```

# 5.Weather Data Analysis


 Analyze historical weather data to identify temperature, rainfall, and seasonal patterns.

***Load weather data using Pandas***

```python
df=pd.read_csv("weather_data_150_rows.csv")
```

```python
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.columns
```

***Handle missing weather observations***.


```python
print(df.isnull().sum())

df['Temperature_C']=df['Temperature_C'].fillna(
    df['Temperature_C'].mean()
)
df['Rainfall_mm']=df['Rainfall_mm'].fillna(
    df['Rainfall_mm'].mean()
)
df['Humidity_%']=df['Humidity_%'].fillna(
    df['Humidity_%'].mean()
)
df['Wind_Speed_kmh']=df['Wind_Speed_kmh'].fillna(
    df['Wind_Speed_kmh'].mean()
)
print(df.isnull().sum())
```

***Calculate average, minimum, and maximum temperatures using NumPy***

```python
avg_temp = np.mean(df['Temperature_C'])
min_temp = np.min(df['Temperature_C'])
max_temp = np.max(df['Temperature_C'])

print("Average Temprature : ",avg_temp)
print("Minimum Temprature : ",min_temp)
print("Maximum Temprature : ",max_temp)
```

***Analyze monthly and seasonal temperature patterns.***

```python
df['Date'] = pd.to_datetime(df['Date'])

df['Month'] = df['Date'].dt.month

monthly_temperature = df.groupby('Month')['Temperature_C'].mean()

print(monthly_temperature)
```

```python
# For Seasons 
df['Date'] = pd.to_datetime(df['Date'])

df['Month'] = df['Date'].dt.month

monthly_temperature = df.groupby('Month')['Temperature_C'].mean()

print(monthly_temperature)
def get_season(month):
    if month in [1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Autumn"

df['Season'] = df['Month'].apply(get_season)

seasonal_temperature = df.groupby('Season')['Temperature_C'].mean()
    
```

***Find days with extreme temperatures or rainfall.***

```python
# Hottest day
hottest_day = df.loc[df['Temperature_C'].idxmax()]

#Coldest day
coldest_day = df.loc[df['Temperature_C'].idxmin()]

# Highest rainfall day 
wettest_day = df.loc[df['Rainfall_mm'].idxmax()]

print("Hottest Day :")
print(hottest_day[["Date","Temperature_C"]])

print("\nColdest Day :")
print(coldest_day[["Date","Temperature_C"]])

print("\nWettest Day : ")
print(wettest_day[["Date","Rainfall_mm"]])
```

# Create:


***1. Temperature trend line chart.***

```python
plt.figure(figsize=(12,5))
plt.plot(df["Date"],df["Temperature_C"],marker='o',color='yellow')
plt.title("Temprature Trends")
plt.xlabel("Date")
plt.ylabel("Temperature(°C)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

***Monthly rainfall bar chart***

```python
monthly_rainfall = df.groupby(df["Date"].dt.month_name())["Rainfall_mm"].sum()
plt.figure(figsize=(10,5))
plt.bar(monthly_rainfall.index,monthly_rainfall.values)
plt.title("Monthly Rainfall")
plt.xlabel("Months")
plt.ylabel("Rainfall(mm)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

***Temperature distribution histogram***

```python
plt.figure(figsize=(10,5))
plt.hist(df["Temperature_C"],bins=10)
plt.title("Temprature Distrubution")
plt.xlabel("Temprature(°C)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

***Box plots for seasonal temperatures***

```python
#Convert Date
df["Date"] = pd.to_datetime(df["Date"])

# Create Season
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Autumn"

df["Season"] = df["Date"].dt.month.map(get_season)

# Box plot
plt.figure(figsize=(10, 5))

df.boxplot(column="Temperature_C", by="Season")

plt.title("Seasonal Temperature Distribution")
plt.suptitle("")
plt.xlabel("Season")
plt.ylabel("Temperature (°C)")

plt.tight_layout()
plt.show()
```

***Seaborn heatmap for weather-variable correlations***

```python
weather_data = df[[
    "Temperature_C",
    "Rainfall_mm",
    "Humidity_%",
    "Wind_Speed_kmh"
]]

correlation = weather_data.corr()

plt.figure(figsize=(8,5))

sns.heatmap(correlation,annot=True,cmap="coolwarm",fmt=".2f")

plt.title("Weather Variable Correlation Heatmap")
plt.tight_layout()
plt.show()
```

***Identify the hottest, coldest, and wettest periods.***

```python
# Convert Date
df["Date"] = pd.to_datetime(df["Date"])

# Hottest period
hottest = df.loc[df["Temperature_C"].idxmax()]

# Coolest period
coolest = df.loc[df["Temperature_C"].idxmin()]

# Wettest period
wettest = df.loc[df["Rainfall_mm"].idxmax()]

print("Hottest Period:")
print(hottest["Date"], "-", hottest["Temperature_C"], "°C")

print("\nCoolest Period:")
print(coolest["Date"], "-", coolest["Temperature_C"], "°C")

print("\nWettest Period:")
print(wettest["Date"], "-", wettest["Rainfall_mm"], "mm")
```
