***#Data Cleaning***


***#Create a copy - Always create a copy before major cleaning.***

```python
df_clean=df.copy()
```

```python
df_clean
```

***#Clean age***
First inspect invalid age


```python
df_clean[
    (df_clean['Age']<0)|
    (df_clean['Age']>100)
]
```

```python
# We will consider ages between 18 and 100 as valid for this project 
```

***#Convert invalid ages to missing values***

```python
df_clean.loc[
    (df_clean['Age']<18)|
    (df_clean['Age']>100),
    'Age'
] = np.nan
```

***#Fill missing ages with median.***

```python
df_clean['Age'] = df_clean['Age'].fillna(
    df_clean['Age'].median()
)
```

```python
df_clean['Age'].describe()
```

***#Clean Quantity***

```python
df_clean[
    df_clean["Quantity"]<=0
]
```

***#Convert invalid values to missing***

```python
df_clean.loc[
    df_clean["Quantity"]<=0,
    "Quantity"
] =np.nan
```

***#Fill Using meadian***

```python
df_clean["Quantity"]=df_clean[
    "Quantity"
].fillna(
    df_clean["Quantity"].median()
)
```

***Clean Discount Percent***

#Check invalid discount

```python
df_clean[
    (df_clean["Discount_Percent"]<0)|
    (df_clean["Discount_Percent"]>100)
]
```

***#Convert invalid values***

#Fill Missing values using median

```python
df_clean["Discount_Percent"]=df_clean[
    "Discount_Percent"
].fillna(
    df_clean["Discount_Percent"].median()
)
```

***#Fill Missing Categorical columns***

***#Check Categorical columns***

```python
categorical_columns=df_clean.select_dtypes(
    include="object"
).columns
categorical_columns
```

***#we fill missing categorical values using "Unknown"***

```python
categorical_missing_columns = [
    "Customer_Name",
    "Gender",
    "Customer_Segment",
    "State",
    "City",
    "Sub_Category",
    "Payment_Method"
]

for column in categorical_missing_columns:
    df_clean[column] =df_clean[
        column
    ].fillna("Unknown")
```

***#Convert Date Columns*** 


```python
df_clean["Order_Date"] = pd.to_datetime(
    df_clean["Order_Date"]
)
```

```python
df_clean["Order_Date"]=pd.to_datetime(
    df_clean["Order_Date"],
    errors="coerce"
)
```

```python
df_clean["Delivery_Date"]=pd.to_datetime(
    df_clean["Delivery_Date"]
)
```

***#Create Delivery Days***

```python
df_clean["Delivery_Days"] = (
    df_clean["Delivery_Date"] -
    df_clean["Order_Date"]
).dt.days
```

# Feature Engineering 

```python
df_clean["Order_Year"] = (
    df_clean["Order_Date"].dt.year
)
```

```python
df_clean["Order_Year"] = (
    df_clean["Order_Date"].dt.year
)
```

```python
df_clean["Month_Number"] = (
    df_clean["Order_Date"].dt.month
)
```

```python
# Profit Margin 
# Profit Margin = 
# Profit / Sales x 100
```

```python
df_clean["Profit_Margin"] = (
    df_clean["Profit"] /
    df_clean["Sales"]
) * 100
```

```python
# Create Age Group
```

```python
df_clean["Age_Group"] = pd.cut(
    df_clean["Age"],
    bins=[
        0,
        25,
        35,
        45,
        55,
        65,
        100
    ],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56-65",
        "65+"
    ]
)
```
