***Dashboard***

```python
import matplotlib.pyplot as pyplot

months=['Jan','Feb','Mar','Apr','May','Jun']

sales=[120,150,180,140,200,250]
expenses=[80,100,120,90,130,110]
```

```python
fig,ax=plt.subplots(2,2,figsize=(10,7))
# Sales Line Chart

ax[0,0].plot(months,sales,marker='o',color="pink")
ax[0,0].set_title("Monthly Sales",color="yellow")
ax[0,0].set_xlabel("Month")
ax[0,0].set_ylabel("Sales")


#2 Expenses Bar Chart

ax[0,1].bar(months,expenses,color="pink")
ax[0,1].set_title("Monthly Expenses",color="yellow")
ax[0,1].set_xlabel("Months")
ax[0,1].set_ylabel("Expenses")


#3 Sales V/S Expenses

ax[1,0].plot(months,sales,label="Sales",color="pink")
ax[1,0].plot(months,expenses,label="Expenses",color="yellow")
ax[1,0].set_title("Sales v/s Expenses",color="yellow")
ax[1,0].legend()


#4.Sales Pie Chart

ax[1,1].pie(sales,labels=months,autopct='%1.1f%%')
ax[1,1].set_title("Sales Distribution",color="yellow")

fig.suptitle("Business Dashboard",fontsize=16,color="red")
plt.tight_layout(pad=2)
plt.subplots_adjust(hspace=0.6)
plt.show()

```
