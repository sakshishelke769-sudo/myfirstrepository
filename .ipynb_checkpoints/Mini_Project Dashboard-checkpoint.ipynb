***Visualization***

```python
city_price = df.groupby("City")["Tour_Price_INR"].mean().sort_values()
plt.figure(figsize=(10,6))
plt.barh(city_price.index,city_price.values)
plt.title("Average Tour price by City")
plt.xlabel("Average Tour price (INR)")
plt.ylabel("City")
plt.show()
```

```python
sns.boxenplot(
    data=df,
    x="Tour_Type",
    y="Tour_Price_INR"
)

plt.title("Tour Price Distribution By Tour Type")
plt.show()
```

```python
# City Booking Chart
```

```python
city_booking= (
    df.groupby("City")["Bookings"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

sns.barplot(
    x=city_booking.values,
    y=city_booking.index
)

plt.title("Top 10 Cities by Booking")
plt.xlabel("Total Bookings")
plt.ylabel("City")
plt.show()
```

```python
# Corelation Analysis 
```

```python
numeric_colums = [
    "Duration_Days",
    "Tour_Price_INR",
    "Travelers",
    "Rating",
    "Bookings",
    "Hotel_Cost",
    "Flight_Cost",
    "Food_Cost",
    "Activities_Cost",
    "Local_Transport_Cost"
          
]
corr = df[numeric_colums].corr()
plt.figure(figsize=(12,8))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.title("Corelation Matrix")
plt.show()
```
