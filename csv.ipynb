```python
#What are the top 5 most populated cities?


df[df["population"]==df["population"].max()]
```

```python
#which city has the higest population ?

df[df["population"]==df["population"].max()]["city"]
```

```python
#which city has the lowest population ?

df[df["population"]==df["population"].min()]["city"]
```

```python
#which country contains the city with the highest population?

df[df["population"]==df["population"].max()]["country"]
```

```python
#which state/region contains the most populated city?

df[df["population"]==df["population"].max()]["state"]
```
