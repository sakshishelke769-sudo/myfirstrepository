```python
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(16,10))

#plot 1 : Gender count
plt.subplot(2,2,1)
sns.countplot(data=df,x="gender")
plt.title("Number of students by Gender")

#Plot 2 : Class count

plt.subplot(2,2,2)
sns.countplot(data=df,x="class")
plt.title("Number of student in each class ")

# plot 3 : average score by class 
plt.subplot(2,2,3)
sns.barplot(
    data=df,
    x="class",
    y="average"
)
plt.title("Average score by class")

# Plot 4 : Attendance Distribution 

plt.subplot(2,2,4)
sns.histplot(
    data=df,
    x="attendance",
    bins=10,
    kde=True
)

plt.title("Attendance Distribution")

# Adjust Spacing
plt.tight_layout()

#Show Dashbord
plt.show()
```
